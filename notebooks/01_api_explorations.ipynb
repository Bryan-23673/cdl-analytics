# API Exploration

## Setup
- Load Libraries
- Load .env
- Verify API key loaded

```python
import os
import requests
import pandas as pd

from dotenv import load_dotenv

load_dotenv()

CITO_API_KEY = os.getenv("CITO_API_KEY")

if os.getenv("CITO_API_KEY"):
    print("API key loaded successfully")
else:
    print("API key not found or is empty")
```

## Teams
- Organizations endpoint
- Team stats endpoint
- Recent matches endpoint

```python
import json 
organizations_and_teams_url = "https://api.citoapi.com/api/v1/cod/orgs?franchiseOnly=true&limit=12"

headers = {
    "x-api-key" : CITO_API_KEY
}

response = requests.get(organizations_and_teams_url, headers=headers)

print(response.status_code)
print(json.dumps(response.json(), indent=4))
```

## Organizations and Teams Endpoint

### Purpose
Returns all Call of Duty League organizations (teams).

### Endpoint
GET /api/v1/cod/orgs

### Important Fields
- slug
- name
- shortName
- logoUrl
- region
- city
- wikiUrl
- approxTotalWinnings

### Possible Uses
- Display all CDL teams
- Team search
- Team logos
- Team filters
- Link to team profile page

### Notes
- `slug` appears to be the unique identifier.
- `logoUrl` can be used to display team logos.
- `approxTotalWinnings` could be displayed on the team page.
- `wikiUrl` could be used as an external reference.

```python
team_stats_url = "https://api.citoapi.com/api/v1/cod/teams/optic-texas/stats"

response = requests.get(team_stats_url, headers=headers)

print(response.status_code)
print(json.dumps(response.json(), indent=4))
```

## Team Stats Endpoint

### Purpose
Returns all stats of a chosen Call of Duty League organization.

### Endpoint
GET /api/v1/cod/teams/{slug}/stats

### Important Fields
- organization
- filter
- totalMapsPlayed
- totalMatches
- byMode
- byMap
- bestMaps
- worstMaps

### Possible Uses
- Team overview page
- Compare performance across game modes
- Display map records and win rates
- Find cmoonly picked and banned maps
- Compare average points scored and allowed
- Identify team strengths and weaknesses

### Notes
- Win rate should be shown together with maps played because small samples can be misleading
- Each map may support different game modes.
- The `byMap` section contains more detailed information than `bestMaps` and `worstMaps`.
- The API may apply a minimum sample size when calculating best and worst maps.
- The season is incldued in the response and can later be used as a dashboard filter

```python
team_recent_matches_url = "https://api.citoapi.com/api/v1/cod/teams/optic-texas/recent-matches"

response = requests.get(team_recent_matches_url, headers=headers)

print(response.status_code)
print(json.dumps(response.json(), indent=4))
```

## Team Recent Matches Endpoint

### Purpose
Returns the five more recent matches for a selected team.

### Endpoint
GET /api/v1/cod/teams/{slug}/recent-matches

### Important Fields
- organization
- recentMatches
- matchId
- date
- opponent
- result
- tournament
- round

### Possible Uses
- Display recent results on a team page
- Show a team's recent form
- Display opponent logos and series scores
- Link to a match detail page using `matchId`

### Notes
- The endpoint currently returns five matches.
- It is useful for recent form, but not full-season analysis.
- The full. timestamp should be kept because `dateFormatted` may change based on timezone.
- The response does not include map-level or player-level statistics.

```python
player_stats_url = "https://api.citoapi.com/api/v1/cod/players/shotzzy/stats"

response = requests.get(player_stats_url, headers=headers)

print(response.status_code)
print(json.dumps(response.json(), indent=4))
```

# Player Statistics Endpoint

## Purpose

Returns a selected player's current-season statistics and historical season statistics.

## Endpoint

GET /api/v1/cod/players/{id}/stats

## Important Fields

- player
- currentTeam
- currentSeasonStats
- season
- overall
- kills
- deaths
- assists
- kdRatio
- damagePerTen
- hardpoint
- hillTime
- searchAndDestroy
- firstBloods
- aggregate
- historicalStats
- lastUpdated

## Possible Uses

- Display a player's current-season statistics
- Show kills, deaths, assists, and K/D ratio
- Compare players on the same team
- Rank players by K/D, kills, or assists
- Display mode-specific statistics
- Show performance across different seasons
- Add statistics to player cards or player detail pages

## Notes

- The endpoint returns statistics for one player at a time.
- `codPlayerId` connects this response to the roster and player profile endpoints.
- The overall K/D appears consistent with kills divided by deaths.
- Some fields return zero even though the player has recorded activity.
- `mapsPlayed`, `matchesPlayed`, `mapWins`, and `mapLosses` may currently be incomplete.
- Some mode-specific statistics may also be incomplete or unavailable.
- The statistics should be tested across multiple players before being used for rankings.
- `lastUpdated` should be checked because the data may not be updated at the same time as the player profile endpoint.
